```python
import pandas as pd
import matplotlib.pyplot as plt
# For demo purposes, we create a small fictional healthcare.csv if it does not exist
import os

if not os.path.exists('healthcare.csv'):
    import numpy as np
    np.random.seed(42)
    demo_data = {
        'Appointment_ID': range(1, 21),
        'Patient_ID': [f'P{100+i}' for i in range(20)],
        'Age': np.random.randint(18, 80, 20),
        'Department': np.random.choice(['Cardiology', 'Pediatrics', 'Orthopedics', 'General Medicine'], 20),
        'Appointment_Status': np.random.choice(['Completed', 'Cancelled', 'Missed'], 20, p=[0.6, 0.2, 0.2]),
        'Waiting_Time': np.random.randint(5, 60, 20)
    }
    pd.DataFrame(demo_data).to_csv('healthcare.csv', index=False)

df = pd.read_csv('healthcare.csv')
display(df.head())
print('Dataset shape:', df.shape)
print('\nDataset information:')
print(df.info())
print('\nStatistical summary:')
print(df.describe())
print('Missing values:')
print(df.isnull().sum())
print('\nDuplicate rows:', df.duplicated().sum())
df = df.drop_duplicates()
status_counts = df['Appointment_Status'].value_counts()
print(status_counts)

completed = (df['Appointment_Status'] == 'Completed').sum()
cancelled = (df['Appointment_Status'] == 'Cancelled').sum()
missed = (df['Appointment_Status'] == 'Missed').sum()
total = len(df)
missed_rate = missed / total * 100 if total else 0

print('\nTotal appointments:', total)
print('Completed:', completed)
print('Cancelled:', cancelled)
print('Missed:', missed)
print('Missed appointment rate:', round(missed_rate, 2), '%')
print('Appointments by department:')
print(df['Department'].value_counts())

print('\nMissed appointments by department:')
missed_by_department = df[df['Appointment_Status'] == 'Missed']
print(missed_by_department['Department'].value_counts())
average_wait = df['Waiting_Time'].mean()
median_wait = df['Waiting_Time'].median()
print('Average waiting time:', round(average_wait, 2), 'minutes')
print('Median waiting time:', round(median_wait, 2), 'minutes')

print('\nAverage waiting time by department:')
print(df.groupby('Department')['Waiting_Time'].mean())
def age_group(age):
    if age < 30:
        return 'Young'
    elif age < 50:
        return 'Adult'
    else:
        return 'Senior'

df['Age_Group'] = df['Age'].apply(age_group)
print(df['Age_Group'].value_counts())
print('========== HEALTHCARE DATA ANALYSIS ==========')
print('Total appointments:', total)
print('Completed appointments:', completed)
print('Cancelled appointments:', cancelled)
print('Missed appointments:', missed)
print('Missed appointment rate:', round(missed_rate, 2), '%')
print('Average waiting time:', round(average_wait, 2), 'minutes')
print('Department with most appointments:', df['Department'].value_counts().idxmax())
print('===============================================')
```
